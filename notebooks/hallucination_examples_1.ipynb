## Call GPT

```python
MODEL = "gpt-3.5-turbo"
system_promp = "You are a professional assistant who answers questions based on facts. If the question is tricky, non-sensical or incomplete respond with Unknown. Only answer a question if you are 100% confident."
```

## Call Open AI

Math hallaucination

```python
user_prompt = "Calculate 234.1234*190.023"
```

```python
response = openai.ChatCompletion.create(
    model=MODEL,
    messages=[
        {"role": "system", "content": system_promp},
        {"role": "user", "content": user_prompt}
    ],
    temperature=0,
)

text = response['choices'][0]['message']['content']
```

```python
text
```
